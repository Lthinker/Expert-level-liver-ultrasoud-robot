Input visualization (on phantom)

```python
import pickle
import numpy as np
from matplotlib import pyplot as plt 
with open('./InputState/PreviousState.pkl','rb') as f:
    data = pickle.load(f)
loadframes = np.array(data['frames'])
loadforces = np.array(data['force'])
loadposes = np.array(data['arm_pos'])
plt.figure()
for ii in range(0,loadframes.shape[0]):
    # selective show, but all images will be input
    plt.subplot(6,6,ii+1).imshow(loadframes[ii])
    plt.axis('off')
plt.savefig('figs/input_image.png')
import matplotlib.pyplot as plt
import numpy as np
import matplotlib as mpl

fig = plt.figure(figsize=(16, 12))
fig.suptitle('Previous action in robot space', fontsize=16, fontweight='bold')

titles = ['Fx', 'Fy', 'Fz', 'Tx', 'Ty', 'Tz', 'x', 'y', 'z', 'Rx', 'Ry', 'Rz']

plt.subplots_adjust(hspace=0.4, wspace=0.3)
pre_action = np.concatenate([loadforces,loadposes],axis=-1)
for ii in range(12):
    ax = plt.subplot(4, 3, ii+1)
    ax.plot(pre_action[:, ii], linewidth=2)
    ax.set_title(titles[ii], fontsize=14, fontweight='bold')
    ax.grid(True, linestyle='--', alpha=0.7)
    
    ax.set_xlabel('Time steps', fontsize=10)
    
    if ii < 3:  # Fx, Fy, Fz
        ax.set_ylabel('Force (N)', fontsize=10)
    elif ii < 6:  # Tx, Ty, Tz
        ax.set_ylabel('Torque (Nm)', fontsize=10)
    elif ii < 9:  # x, y, z
        ax.set_ylabel('Position (m)', fontsize=10)
    else:  # Rx, Ry, Rz
        ax.set_ylabel('Rotation (rad)', fontsize=10)
    
    ax.tick_params(labelsize=9)
    
plt.tight_layout()
plt.subplots_adjust(top=0.92)  
plt.savefig('figs/input_poseforce.png')
plt.show()
```

Output visualization

```python
import numpy as np
import os 
from matplotlib import pyplot as plt 
from scipy.spatial.transform import Rotation as R

def rotation_vector_to_euler(rot_vec):
    original_shape = rot_vec.shape
    
    if len(original_shape) == 1:  
        r = R.from_rotvec(rot_vec)
        # (xyz，roll, pitch, yaw)
        euler = r.as_euler('xyz')
        return euler
    elif len(original_shape) == 2: 
        r = R.from_rotvec(rot_vec)
        # (xyz，roll, pitch, yaw)
        euler = r.as_euler('xyz')
        return euler
    else:
        raise ValueError("Wrong input dimensions")
    
fpPredAction = os.path.join('Outputaction/PredAction.npy')
fpExecAction = os.path.join('Outputaction/ExeAction.npy')

Pred = np.load(fpPredAction)
Exec = np.load(fpExecAction)
print(Pred.shape)
print(Exec.shape)

# Convert rot vec to euler angel for better visualization
Exec[:,9:] = rotation_vector_to_euler(Exec[:,9:])
```

```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib as mpl

fig = plt.figure(figsize=(16, 12))
fig.suptitle('Predicted action in harmonized space (Estimate 8 actions)', fontsize=16, fontweight='bold')

titles = ['Fx', 'Fy', 'Fz', 'Tx', 'Ty', 'Tz', 'x', 'y', 'z', 'Rx', 'Ry', 'Rz']

plt.subplots_adjust(hspace=0.4, wspace=0.3)

for ii in range(12):
    ax = plt.subplot(4, 3, ii+1)
    ax.plot(Pred[:, ii], linewidth=2)
    ax.set_title(titles[ii], fontsize=14, fontweight='bold')
    ax.grid(True, linestyle='--', alpha=0.7)
    
    ax.set_xlabel('Time steps', fontsize=10)
    
    if ii < 3:  # Fx, Fy, Fz
        ax.set_ylabel('Force (N)', fontsize=10)
    elif ii < 6:  # Tx, Ty, Tz
        ax.set_ylabel('Torque (Nm)', fontsize=10)
    elif ii < 9:  # x, y, z
        ax.set_ylabel('Position (m)', fontsize=10)
    else:  # Rx, Ry, Rz
        ax.set_ylabel('Rotation (rad)', fontsize=10)
    
    # 让坐标轴更清晰
    ax.tick_params(labelsize=9)
    
plt.tight_layout()
plt.subplots_adjust(top=0.92)  
plt.savefig('figs/output_predposeforce.png')
plt.show()
```
